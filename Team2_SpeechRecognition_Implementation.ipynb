# **3. Speech-to-Text Conversion**

```python
model = whisper.load_model("small")

decode_opts = dict(
    temperature=0,
    beam_size=5,
    condition_on_previous_text=False,
)

result_1 = model.transcribe(short_audio_path, **decode_opts)
result_2 = model.transcribe(long_audio_path,  **decode_opts)
result_3 = model.transcribe(noisy_audio_path, **decode_opts)

transcribed_text_1 = result_1["text"]
transcribed_text_2 = result_2["text"]
transcribed_text_3 = result_3["text"]
```

```python
print("Short English Transcribed Text:")
print(transcribed_text_1)
print("")
print("Long English Transcribed Text:")
print(transcribed_text_2)
print("")
print("Noisy English Transcribed Text:")
print(transcribed_text_3)
```

**Whisper Small model was used to transcribe English audio files, including short, long, and noisy samples, yielding accurate text transcriptions for all.**

# **4. Language Translation**

```python
translated_arabic_text_1 = GoogleTranslator(source="auto", target="ar").translate(transcribed_text_1)
translated_arabic_text_2 = GoogleTranslator(source="auto", target="ar").translate(transcribed_text_2)
translated_arabic_text_3 = GoogleTranslator(source="auto", target="ar").translate(transcribed_text_3)
```

```python
print("Translated Short Arabic Text:")
print(translated_arabic_text_1)
print("")
print("Translated Long Arabic Text:")
print(translated_arabic_text_2)
print("")
print("Translated Noisy Arabic Text:")
print(translated_arabic_text_3)
```

```python
translated_french_text_1 = GoogleTranslator(source="auto", target="fr").translate(transcribed_text_1)
translated_french_text_2 = GoogleTranslator(source="auto", target="fr").translate(transcribed_text_2)
translated_french_text_3 = GoogleTranslator(source="auto", target="fr").translate(transcribed_text_3)
```

```python
print("Translated Short French Text:")
print(translated_french_text_1)
print("")
print("Translated Long French Text:")
print(translated_french_text_2)
print("")
print("Translated Noisy French Text:")
print(translated_french_text_3)
```

```python
translated_spanish_text_1 = GoogleTranslator(source="auto", target="es").translate(transcribed_text_1)
translated_spanish_text_2 = GoogleTranslator(source="auto", target="es").translate(transcribed_text_2)
translated_spanish_text_3 = GoogleTranslator(source="auto", target="es").translate(transcribed_text_3)
```

```python
print("Translated Short Spanish Text:")
print(translated_spanish_text_1)
print("")
print("Translated Long Spanish Text:")
print(translated_spanish_text_2)
print("")
print("Translated Noisy Spanish Text:")
print(translated_spanish_text_3)
```

# **7. System Latency and Real-Time Capability**

```python
import time

audio_path = noisy_audio_path
target_lang = "ar"

t0 = time.time()

# speech-to-text
text = model.transcribe(audio_path, **decode_opts)["text"]
t1 = time.time()

# translation
translated = GoogleTranslator(source="auto", target=target_lang).translate(text)
t2 = time.time()

# text-to-speech
gTTS(text=translated, lang=target_lang).save("timing_test.mp3")
t3 = time.time()

print("=== English to Arabic ===")
print(f"Speech-to-Text:  {t1 - t0:.2f} seconds")
print(f"Translation:     {t2 - t1:.2f} seconds")
print(f"Text-to-Speech:  {t3 - t2:.2f} seconds")
print(f"Total Pipeline Time:  {t3 - t0:.2f} seconds")
```
